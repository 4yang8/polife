```python
history = [{'role': 'user', 'text': '안녕하세요'},
           {'role': 'counselor', 'text': '안녕하세요. 오늘 상담에 오신 것을 환영합니다. 어떤 이야기를 나누고 싶으신지 말씀해주시면 좋을 것 같아요.'}]

message = "안녕하세요. 요즘 너무 식욕이 없어요."

# url = "http://localhost:8000/chat/chat1"
url = "http://141.223.163.135:8000/chat/chat1"  # nlp4
input = {"sender_id": "user_343", "text": message.strip(), "history": history, "persona": ""}

response = requests.post(url, json=input)
response_json = response.json()
print(response_json)  # 응답 형식 확인

# 'message' 키가 존재하는지 확인하고 처리
message = response_json.get('message', '응답 없음')  # 'message'가 없으면 '응답 없음' 사용

# history 업데이트
history = history + [{"role": "user", "text": message.strip()}, {"role": "counselor", "text": message}]
print(message)  # 챗봇의 응답 메시지 출력
```
